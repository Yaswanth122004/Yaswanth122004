```python
from sklearn.datasets import load_iris
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.metrics import accuracy_score
```

```python
# Load data
iris = load_iris()
X = iris.data
y = iris.target
X
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42)
```

```python
# Create a base Random Forest model
rf = RandomForestClassifier(random_state=42)
```

```python
param_grid = {
    'n_estimators': [50, 100, 200],          # Number of trees
    'max_depth': [None, 5, 10],              # Depth of each tree
    'min_samples_split': [2, 5, 10],         # Min samples needed to split a node
    'min_samples_leaf': [1, 2, 4],           # Min samples needed at a leaf node
    'max_features': ['auto', 'sqrt', 'log2'] # Number of features considered at each split
}
```

```python
# Set up GridSearchCV
grid_search = GridSearchCV(
    estimator=rf,
    param_grid=param_grid,
    cv=5,                 # 5-fold cross-validation
    n_jobs=1,            # Use all CPU cores
    verbose=2             # Show progress
)
```

```python

# Run the grid search on the training data
grid_search.fit(X_train, y_train)
```

```python
# Best parameters found
print("Best parameters found:", grid_search.best_params_)
```

```python
# Use the best model to make predictions on test data
best_rf = grid_search.best_estimator_
y_predict = best_rf.predict(X_test)

custom_test = [[4.7, 3.2, 1.3, 0.1]]
print(best_rf.predict(custom_test))
```

```python
# Evaluate
print("test set accuracy:", accuracy_score(y_test, y_predict))
```

```python
from sklearn.metrics import confusion_matrix
print(confusion_matrix(y_test, y_predict))
```

```python
from sklearn.model_selection import cross_val_score
print(cross_val_score(best_rf,X, y, cv=5))
```
